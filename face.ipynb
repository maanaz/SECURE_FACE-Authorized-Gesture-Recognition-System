```python
path="Images"
mylist=os.listdir(path)
```

```python
imgs=[]
names=[]
 
for img in mylist:
    img_path=os.path.join(path,img)
    # print(img_path)
    current_img=cv2.imread(img_path)
    # print(current_img)
    imgs.append(current_img)
    names.append(img.split('.')[0])
```

```python
def face_encoding(Images):
    encodedList=[]
    for img in imgs:
        img1=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
        face_in_frame=fr.face_locations(img1)
        print(face_in_frame)
        encode=fr.face_encodings(img1,face_in_frame)[0]
        encodedList.append(encode)
    return encodedList
```

```python
encoded_list_known=face_encoding(imgs)
```

```python
import cv2
import numpy as np
import face_recognition as fr

# Try DroidCam camera index
video = cv2.VideoCapture(1)

while True:

    suc, img = video.read()

    if not suc or img is None:
        print("Frame not received")
        continue

    # OpenCV uses BGR, face_recognition expects RGB
    img1 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Find faces
    face_in_frame = fr.face_locations(img1)

    # Get face encodings
    face_encode = fr.face_encodings(img1, face_in_frame)

    for f_enc, f_loc in zip(face_encode, face_in_frame):

        # Compare with registered faces
        matches = fr.compare_faces(
            encoded_list_known,
            f_enc
        )

        # Calculate face distances
        face_d = fr.face_distance(
            encoded_list_known,
            f_enc
        )

        # Find closest registered face
        matchIndex = np.argmin(face_d)

        # Check if closest face actually matches
        if matches[matchIndex]:

            name = names[matchIndex]

            # Face coordinates
            top, right, bottom, left = f_loc

            # Draw rectangle
            cv2.rectangle(
                img,
                (left, top),
                (right, bottom),
                (0, 255, 0),
                2
            )

            # Display name
            cv2.putText(
                img,
                name,
                (left, top - 10),
                cv2.FONT_HERSHEY_COMPLEX,
                0.8,
                (0, 255, 0),
                2
            )

            # Display authorization
            cv2.putText(
                img,
                "AUTHORIZED USER",
                (left, bottom + 30),
                cv2.FONT_HERSHEY_COMPLEX,
                0.7,
                (0, 255, 0),
                2
            )

            print("AUTHORIZED USER:", name)

        else:

            top, right, bottom, left = f_loc

            cv2.rectangle(
                img,
                (left, top),
                (right, bottom),
                (0, 0, 255),
                2
            )

            cv2.putText(
                img,
                "UNAUTHORIZED",
                (left, top - 10),
                cv2.FONT_HERSHEY_COMPLEX,
                0.8,
                (0, 0, 255),
                2
            )

            print("UNAUTHORIZED USER")

    # Show camera
    cv2.imshow("DroidCam - Face Authorization", img)

    # Press Q to quit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break


video.release()
cv2.destroyAllWindows()
```

```python


DATASET_PATH = "gesture_dataset"

IMG_SIZE = (224, 224)
BATCH_SIZE = 16
SEED = 42

train_dataset = tf.keras.utils.image_dataset_from_directory(
    DATASET_PATH,
    validation_split=0.2,
    subset="training",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

val_dataset = tf.keras.utils.image_dataset_from_directory(
    DATASET_PATH,
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

class_names = train_dataset.class_names

print("Classes:", class_names)
```

```python
import cv2
import numpy as np
import face_recognition as fr
import tensorflow as tf


gesture_model = tf.keras.models.load_model("gesture_model.keras")
class_names = ['open_palm', 'peace', 'thumbs_up']

video = cv2.VideoCapture(1)

while True:

    success, img = video.read()

    if not success or img is None:
        print("Frame not received")
        continue


    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Make sure dlib receives a valid 8-bit contiguous RGB image
    img_rgb = np.ascontiguousarray(img_rgb, dtype=np.uint8)

    face_locations = fr.face_locations(img_rgb)
    face_encodings = fr.face_encodings(img_rgb, face_locations)
    authorized = False

    for face_encode, face_loc in zip(face_encodings, face_locations):

        matches = fr.compare_faces(
            encoded_list_known,
            face_encode
        )

        face_distances = fr.face_distance(
            encoded_list_known,
            face_encode
        )

        match_index = np.argmin(face_distances)

        top, right, bottom, left = face_loc

        if matches[match_index]:

            name = names[match_index]
            authorized = True

        
            cv2.rectangle(
                img,
                (left, top),
                (right, bottom),
                (0, 255, 0),
                2
            )

            cv2.putText(
                img,
                name,
                (left, top - 10),
                cv2.FONT_HERSHEY_COMPLEX,
                0.8,
                (0, 255, 0),
                2
            )

            cv2.putText(
                img,
                "AUTHORIZED",
                (left, bottom + 30),
                cv2.FONT_HERSHEY_COMPLEX,
                0.7,
                (0, 255, 0),
                2
            )

        else:

            # Red face box
            cv2.rectangle(
                img,
                (left, top),
                (right, bottom),
                (0, 0, 255),
                2
            )

            cv2.putText(
                img,
                "UNAUTHORIZED",
                (left, top - 10),
                cv2.FONT_HERSHEY_COMPLEX,
                0.8,
                (0, 0, 255),
                2
            )

    if authorized:
        gesture_img = cv2.resize(img, (224, 224))


        gesture_img = cv2.cvtColor(
            gesture_img,
            cv2.COLOR_BGR2RGB
        )

        # Convert to batch
        gesture_img = np.expand_dims(
            gesture_img,
            axis=0
        )

        # MobileNetV2 preprocessing
        gesture_img = tf.keras.applications.mobilenet_v2.preprocess_input(
            gesture_img
        )

        # Prediction
        predictions = gesture_model.predict(
            gesture_img,
            verbose=0
        )

        predicted_index = np.argmax(predictions[0])

        gesture = class_names[predicted_index]

        confidence = predictions[0][predicted_index]

        # Display gesture
        cv2.putText(
            img,
            f"Gesture: {gesture}",
            (20, 50),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.9,
            (0, 255, 0),
            2
        )

        cv2.putText(
            img,
            f"Confidence: {confidence * 100:.1f}%",
            (20, 85),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (0, 255, 0),
            2
        )

    else:

        cv2.putText(
            img,
            "GESTURE ACCESS BLOCKED",
            (20, 50),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (0, 0, 255),
            2
        )

    # Show camera
    cv2.imshow(
        "Secure Gesture Recognition",
        img
    )

    # Press Q to quit
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break


video.release()
cv2.destroyAllWindows()
```

```python
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

def preprocess(image, label):
    image = preprocess_input(image)
    return image, label
```

```python
train_preprocessed = train_dataset.map(preprocess)
val_preprocessed = val_dataset.map(preprocess)
```

```python
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras import layers, models

base_model = MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

base_model.trainable = False

model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.2),
    layers.Dense(3, activation="softmax")
])
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
history = model.fit(
    train_preprocessed,
    validation_data=val_preprocessed,
    epochs=10
)
```
